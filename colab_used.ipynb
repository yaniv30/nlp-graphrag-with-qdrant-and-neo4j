```python
import re

def extract_triplets(log_path):
    triplets = []
    pattern = re.compile(r"Processed triplet: \((.*?)\)")

    with open(log_path, 'r', encoding='utf-8') as f:
        for line in f:
            match = pattern.search(line)
            if match:
                raw_triplet = match.group(1)
                triplet = tuple(s.strip().strip("'") for s in raw_triplet.split(","))
                triplets.append(triplet)
    return triplets

def save_triplets_to_file(triplets, output_path):
    with open(output_path, 'w', encoding='utf-8') as f:
        for triplet in triplets:
            f.write(f"{triplet}\n")

# Utilisation
log_file = '/content/nlp-graphrag-with-qdrant-and-neo4j/logs/graphrag.log'
output_file = 'triplets.txt'

triplets = extract_triplets(log_file)
save_triplets_to_file(triplets, output_file)

from google.colab import files
files.download(output_file)
```
